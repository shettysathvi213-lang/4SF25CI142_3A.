```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df.drop_duplicates(inplace=True)
```

```python

df['rate'] = df['rate'].str.split('/').str[0]
print(df['rate'])
```

```python
print(df['rate'].unique())
```

```python
df['rate'] = df['rate'].replace( ['NEW', '-', 'nan', 'None', 'NaN'], np.nan)
```

```python
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
print(df['rate'])
```

```python
df['approx_cost(for two people)'] = ( df['approx_cost(for two people)'].astype(str).str.replace(',', '', regex=False))
```

```python
df['approx_cost(for two people)'] = pd.to_numeric( df['approx_cost(for two people)'])
```

```python
print(df['approx_cost(for two people)'])
```

```python
df['votes'] = pd.to_numeric(df['votes'], errors='coerce')
print(df['votes'])
```

```python
df.info()
```

```python
#             total number of null values
df.isnull().sum().sum()
```

```python
df.shape
```

```python
df.dropna().shape
```

```python
df.dropna(axis=1).shape
```

```python
df['rate'].mean()
```

```python
df['rate_mean'] = df['rate'].fillna(df['rate'].mean())

df[['rate', 'rate_mean']].head(10)
```

```python
df['votes_mean'] = df['votes'].fillna(df['votes'].mean())
```

```python
df['cost_median'] = df['approx_cost(for two people)'].fillna( df['approx_cost(for two people)'].median())
df[['cost_median','approx_cost(for two people)']].head(10)
```

```python
df['online_order'].mode()
```

```python
df['online_order'] = df['online_order'].fillna(df['online_order'].mode()[0])
```

```python
df['cuisines'] = df['cuisines'].fillna(df['cuisines'].mode()[0])
```

```python
df['book_table'] = df['book_table'].fillna( df['book_table'].mode()[0])
```

```python
df.isnull().sum()
```

```python

top_locations = df['location'].value_counts().head(10)

print(top_locations)
```

```python
def rating_category(rating):
    if rating >= 4:
        return 'Excellent'
    elif rating >= 3:
        return 'Good'
    elif rating >= 2:
        return 'Average'
    else:
        return 'Poor'

df['rating_category'] = df['rate'].apply(rating_category)
```

```python
df['cost_per_person'] = df['approx_cost(for two people)'] / 2

df[['approx_cost(for two people)', 'cost_per_person']].head(10)
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='votes',
    y='rate',
    data=df,
    color='green',
)

plt.title('Scatter Plot of Votes vs Ratings')
plt.xlabel('Number of Votes')
plt.ylabel('Restaurant Rating')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.lineplot(
    x='rating_category',
    y='approx_cost(for two people)',
    data=df
)

plt.title('Average Cost by Rating Category')
plt.xlabel('Rating Category')
plt.ylabel('Approximate Cost for Two')
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    x='online_order',
    y='rate',
    data=df
)

plt.title('Online Ordering vs Average Rating')
plt.xlabel('Online Ordering')
plt.ylabel('Average Rating')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='online_order',
    y='approx_cost(for two people)',
    data=df
)

plt.title('Cost Distribution by Online Ordering')
plt.xlabel('Online Ordering')
plt.ylabel('Approximate Cost for Two')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.violinplot(
    x='book_table',
    y='rate',
    data=df
)

plt.title('Rating Distribution by Table Booking')
plt.xlabel('Table Booking')
plt.ylabel('Restaurant Rating')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['approx_cost(for two people)'],
    bins=15,
    kde=True
)

plt.title('Distribution of Restaurant Costs')
plt.xlabel('Approximate Cost for Two')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[['rate', 'votes',
        'approx_cost(for two people)',
        'cost_per_person']].corr(),
)

plt.title("Correlation Heatmap")
plt.show()
```
